#### 15. 利用13题目中的x，y 输出 x*y ,和 np.multiply(x, y) 还有  np.dot(x,y),比较差异。然后自己换一个不是方阵的试试。

```python
print("x*y：", x*y)
print("np.multiply(x,y)：", np.multiply(x,y))
print("np.dot(x,y)：", np.dot(x,y))

# 换一个不是方阵的
x2 = np.array([[1, 2], [3, 4], [5, 6]])
y2 = np.array([[6, 5], [4, 3], [2, 1]])

print("===== 不是方阵的例子 =====")

try:
    print(x2*y2)
except ValueError:
    print("无法进行 x2*y2")
    
try:
    print(np.multiply(x2,y2))
except ValueError:
    print("无法进行 np.multiply(x,y)")
    
try:
    print(np.dot(x2,y2))
except ValueError:
    print("无法进行 np.dot(x,y)")
```

#### 24,画图，y=x*x 其中 x = np.arange(0, 100, 0.1) （提示这里用到  matplotlib.pyplot 库）

```python
import matplotlib.pyplot as plt

x = np.arange(0, 100, 0.1)
y = x*x

plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle="--", alpha=0.6)

plt.plot(x, y)
plt.show()
```

#### 25.画图。画正弦函数和余弦函数， x = np.arange(0, 3 * np.pi, 0.1)(提示：这里用到 np.sin() np.cos() 函数和 matplotlib.pyplot 库)

```python
x = np.arange(0, 3*np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.xlabel("x")
plt.ylabel("y")

plt.plot(x, y_sin, label=r'$\sin(x)$', color='b', linewidth=2)
plt.plot(x, y_cos, label=r'$\cos(x)$', color='r', linewidth=2)

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
